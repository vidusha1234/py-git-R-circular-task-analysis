# Comparison and validation of the two circular task datasets

## 1. Objective

The objective of this analysis is to compare the professor's reference dataset with our own MouseReMoCo recording.

The comparison focuses on the acquisition conditions, the number of recorded samples, the effective sampling frequency, the duration of each recording and the performance indicators obtained from the circular task.

The aim is also to identify possible differences between the two datasets and determine whether these differences may be related to the acquisition conditions or to the analysis procedure.

## 2. Comparison of acquisition conditions

The two datasets were recorded using different software versions and screen configurations.

| Parameter | Professor dataset | Personal dataset |
|---|---|---|
| Software | LSL-mouse | MouseReMoCo |
| Version | 1.2.0rc5 | 1.3.0 |
| Screen resolution | 1104 × 661 | 1350 × 689 |
| Screen centre | (552, 330) | (675, 344) |
| Task radius | 209.5 pixels | 223.5 pixels |
| Task tolerance | 47 pixels | 47 pixels |
| Configured cycles | 6 | 6 |
| Cycle duration | 20 seconds | 20 seconds |

Although both tasks were configured with six cycles of 20 seconds, only five complete recording periods were found in each marker file.

The two datasets therefore contain five usable recordings each.

## 3. Effective sampling frequency

One of the main differences between the datasets concerns the effective sampling frequency.

For each recording, the frequency was calculated using the number of recorded samples divided by the duration of the recorded data.

| Recording | Professor dataset (Hz) | Personal dataset (Hz) |
|---|---:|---:|
| 1 | 198.68 | 27.77 |
| 2 | 198.94 | 22.83 |
| 3 | 198.03 | 21.30 |
| 4 | 199.53 | 23.18 |
| 5 | 197.22 | 23.90 |

The professor's dataset has an effective sampling frequency close to 200 Hz, whereas our personal dataset has a frequency between approximately 21 and 28 Hz.

This represents a substantial difference in the temporal resolution of the recordings.

A lower sampling frequency means that fewer cursor positions are available to reconstruct the movement trajectory. This may affect the precision of trajectory analysis and the calculation of performance indicators.

However, the exact cause of this difference cannot be determined from the available data alone. It may be related to the software version, the computer, the mouse or the recording conditions.

## 4. Recording duration and marker analysis

The marker files indicate that each recording period was designed to last approximately 20 seconds.

However, the duration covered by the actual position samples differs between the two datasets.

For the professor's dataset, the recorded samples cover approximately 19.5 to 20 seconds per cycle.

For our personal dataset, the recorded samples cover approximately 16.7 to 18.9 seconds per cycle.

This difference suggests that the available position samples do not cover the same proportion of the marker-defined recording periods.

Further investigation would be needed to determine whether this is related to the way mouse positions are recorded or to the acquisition process.

The marker analysis also confirms that only five complete recording cycles are available in each dataset, despite the six cycles configured in the software.

## 5. Comparison of performance indicators

The same statistical function was applied to both datasets. The mean values across the five recordings are presented below.

| Indicator | Professor dataset | Personal dataset |
|---|---:|---:|
| Absolute number of laps | 12.73 | 1.19 |
| Mean radius (Re) | 210.70 px | 222.46 px |
| Radial variability (Te) | 49.94 px | 23.43 px |
| Error rate | 2.39% | 0.00% |
| Movement time per lap (MT) | 1.53 s | 15.32 s |
| Effective index of difficulty (IDe) | 29.91 bits/lap | 61.68 bits/lap |
| Effective bandwidth (Be) | 1.06 | 0.50 |
| Effective information processing (IPe) | 20.06 bits/s | 4.00 bits/s |

The personal dataset presents lower radial variability and a lower error rate than the professor's dataset.

However, the professor's dataset shows a much shorter movement time per lap and a higher effective information processing rate.

These results indicate different movement performances between the two recordings.

It is important to note that the number of laps is negative in the original calculations because it represents the direction of rotation. The comparison table therefore presents its absolute value.

## 6. Verification of the analysis procedure

The processing procedure was checked to identify possible differences in the calculations.

Both notebooks use the same general approach:

- Loading the position data and the marker files.
- Extracting the task configuration from the header.
- Recentring the cursor coordinates around the centre of the screen.
- Identifying recording periods using the `DoRecord` and `DoPause` markers.
- Removing the initial approach phase by starting the analysis at the first sample inside the target.
- Calculating the same performance indicators using the same statistical function.

No major difference was identified in the calculation procedure.

Therefore, the observed discrepancies are more likely to be associated with the acquisition conditions and the characteristics of the recorded datasets than with differences in the statistical code.
